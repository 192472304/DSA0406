```python
import numpy as np

student_scores = np.array([
    [80, 75, 90, 85],
    [70, 80, 85, 75],
    [90, 85, 80, 90],
    [85, 90, 95, 80]
])

avg = np.mean(student_scores, axis=0)
subjects = ["Math", "Science", "English", "History"]

print("Average scores:", avg)
print("Highest average subject:", subjects[np.argmax(avg)])
```

```python
import numpy as np

sales = np.array([
    [100, 200, 300],
    [150, 250, 350],
    [200, 300, 400]
])

avg = np.mean(sales)

print("Average price:", avg)
```

```python
import numpy as np

house_data = np.array([
    [3, 1200, 200000],
    [5, 1800, 350000],
    [6, 2000, 400000],
    [4, 1500, 250000],
    [7, 2200, 500000]
])

houses = house_data[house_data[:, 0] > 4]
avg = np.mean(houses[:, 2])

print("Average sale price:", avg)
```

```python
import numpy as np

sales_data = np.array([10000, 12000, 15000, 18000])

total = np.sum(sales_data)
increase = ((sales_data[3] - sales_data[0]) / sales_data[0]) * 100

print("Total sales:", total)
print("Percentage increase:", increase, "%")
```

```python
import numpy as np

fuel_efficiency = np.array([25, 30, 35, 40])

avg = np.mean(fuel_efficiency)

model1 = fuel_efficiency[0]
model2 = fuel_efficiency[1]

improvement = ((model2 - model1) / model1) * 100

print("Average fuel efficiency:", avg)
print("Percentage improvement:", improvement, "%")
```

```python
import numpy as np

prices = np.array([100, 200, 50])
quantity = np.array([2, 1, 3])

discount_rate = 10
tax_rate = 5

subtotal = np.sum(prices * quantity)
discount = subtotal * discount_rate / 100
after_discount = subtotal - discount
tax = after_discount * tax_rate / 100
total = after_discount + tax

print("Subtotal:", subtotal)
print("Discount:", discount)
print("Tax:", tax)
print("Total cost:", total)
```
